```python
import pyspark
print("Spark version:", pyspark.__version__)
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("day44_latihan") \
    .master("local[*]") \
    .getOrCreate()

sc = spark.sparkContext
print("Master Mode:", sc.master)
print(f"Default Parallelism: {sc.defaultParallelism}")
print("sparkId:", sc.applicationId)
```

```python
import time
rdd_besar = sc.parallelize(range(10000000))  # 10 juta angka
# 1. Transformasi Lazy (Hanya catat rencana)
t0 = time.time()
rdd_genap = rdd_besar.filter(lambda x: x % 2 == 0)
t_lazy = time.time() - t0
print(f"Waktu Transformasi filter() [LAZY] : {t_lazy:.6f} detik (Hampir 0 detik!)")
# 2. Aksi Eager (Baru kompor dinyalakan)
t1 = time.time()
hasil_tiga = rdd_genap.take(3)
t_eager = time.time() - t1
print(f"Waktu Aksi take(3)          [EAGER]: {t_eager:.6f} detik")
print(f"Hasil Ambil 3 Angka Genap          : {hasil_tiga}")
```
